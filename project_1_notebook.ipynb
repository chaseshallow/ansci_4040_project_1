```python
DATA_FILE = "Data_set_prep_assignment_1.csv"

import pandas as pd


def load_data(path=DATA_FILE):
    df = pd.read_csv(path, low_memory=False)
    rename_map = {
        "AnimalId": "animal_number",
        "Avgmilkflow": "average_milk_flow",
        "Flow30_60Session": "30_60_session",
        "YieldFirst2Min_Session": "yield_first_2_min_session_yield",
        "YieldSession": "session_yield",
        "DurationSession_sec": "session_duration",
        "milking": "session_milking",
        "LactationNumber": "lactation_number",
        "DaysInMilk": "days_in_milk",
        "ReproductionStatus": "reproduction_status",
        "EventDate": "event_date",
        
    }
    return df.rename(columns=rename_map)


data = load_data()
data.head()
```

```python
missing_summary = pd.DataFrame({
    'missing_count': data.isna().sum(),
    'missing_pct': (data.isna().mean() * 100).round(2)
})

missing_summary = missing_summary[missing_summary['missing_count'] > 0].sort_values('missing_count', ascending=False)
missing_summary
```

```python
exact_duplicate_cols = [
    'animal_number',
    'event_date',
    'lactation_number',
    'days_in_milk',
    'session_secmilking',
    'avgmilkflow_flow',
    'flow_30_60_session',
    'yield_first_2_min_session_yield',
    'session_yield',
    'session_duration_sec',
    'reproduction_status',
]

exact_duplicates = data[data.duplicated(subset=exact_duplicate_cols, keep=False)]
exact_duplicates = exact_duplicates.sort_values(exact_duplicate_cols)

clean_data = data.drop_duplicates(subset=exact_duplicate_cols, keep='first')

print('duplicate_rows_removed:', len(data) - len(clean_data))
print('rows_after_cleanup:', len(clean_data))
clean_data.head()
```

```python
# Drop rows with NA in average_milk_flow, and drop rows with 5 or more total NA/0 values in the row
base_df = clean_data.copy() if 'clean_data' in globals() else data.copy()

avg_col = 'average_milk_flow' if 'average_milk_flow' in base_df.columns else 'avgmilkflow_flow'

avg_missing = (
    base_df[avg_col].isna()
    | base_df[avg_col].astype(str).str.strip().str.lower().eq('na')
)

row_bad_count = (
    base_df.apply(
        lambda col: col.isna()
        | col.astype(str).str.strip().str.lower().eq('na')
        | col.eq(0)
        | col.astype(str).str.strip().str.lower().eq('0')
    )
).sum(axis=1)

filtered_data = base_df.loc[~avg_missing & (row_bad_count < 5)].copy()

print('rows before dropping:', len(base_df))
print('rows after dropping:', len(filtered_data))
print('rows dropped:', len(base_df) - len(filtered_data))
filtered_data.head()
```

```python
numeric_cols = [
    col for col in clean_data.select_dtypes(include='number').columns
    if col not in ['animal_number', 'lactation_number']
]

outlier_summary = []
for col in numeric_cols:
    s = pd.to_numeric(clean_data[col], errors='coerce')
    q1 = s.quantile(0.25)
    q3 = s.quantile(0.75)
    iqr = q3 - q1
    lower = q1 - 1.5 * iqr
    upper = q3 + 1.5 * iqr
    outliers = s[(s < lower) | (s > upper)]
    if not outliers.empty:
        outlier_summary.append({
            'column': col,
            'lower_bound': lower,
            'upper_bound': upper,
            'count_outliers': int(outliers.count()),
            'min_outlier': float(outliers.min()),
            'max_outlier': float(outliers.max()),
        })

outlier_summary = pd.DataFrame(outlier_summary).sort_values('count_outliers', ascending=False)
outlier_summary
```

```python
import numpy as np

numeric_zscore_cols = [
    col for col in clean_data.select_dtypes(include='number').columns
    if col not in ['animal_number', 'lactation_number']
]

zscore_df = (
    clean_data[numeric_zscore_cols]
    .apply(lambda col: (col - col.mean()) / col.std(ddof=0))
    .replace([np.inf, -np.inf], np.nan)
)

abs_z = zscore_df.abs()
outlier_mask = abs_z > 2
outlier_rows = outlier_mask.any(axis=1)

filtered_by_z = clean_data.loc[~outlier_rows].copy()

print('rows before z-score filtering:', len(clean_data))
print('rows after z-score filtering:', len(filtered_by_z))
print('rows removed by z-score rule:', len(clean_data) - len(filtered_by_z))
print('columns checked:', numeric_zscore_cols)
filtered_by_z.head()
```
